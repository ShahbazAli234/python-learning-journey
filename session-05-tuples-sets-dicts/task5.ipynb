###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
# Write Code here

test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

result = []
processed = []

for t1 in test_list:
    if t1[0] in processed:
        continue
    merged = [t1[0], t1[1]]

    for t2 in test_list:
        if t1 != t2 and t1[0] == t2[0]:
            merged.append(t2[1])
    processed.append(t1[0])
    result.append((tuple(merged)))

print(result)
````

###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
# write your code here

t = (1, 5, 7, 8, 10)

result = []

for i in range(len(t)):
    if i == 0:
        total = t[i] * t[i+1]
    elif i ==  len(t) - 1:
        total = t[i] * t[i-1]
    else:
        total = t[i] * t[i-1] + t[i] * t[i+1]
    result.append(total)

print(tuple(result))
````

###`Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
# write your code here

list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

list_count = 0
set_count = 0
tuple_count = 0

for item in list1:
    if type(item) == list:  # if isinstance(item, list):  --> OOP Concept
        list_count += 1     #    list_count += 1
    elif type(item) == set:
        set_count += 1
    elif type(item) == tuple:
        tuple_count += 1

print("List =",list_count)
print("Sets =",set_count)
print("Tuples =",tuple_count)
````

Given a dictionary with values list, extract key whose value has most unique values.

**Example 1:**

```bash
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```


````python
# write your code here

test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}

max_count = 0

for key in test_dict:
    unique_count = len(set(test_dict[key]))

    if unique_count > max_count:
        max_count = unique_count
        result = key

print(result)
````

### `Q2`: Replace words from Dictionary. Given String, replace it’s words from lookup dictionary.

**Example 1:**

Input:


Output:

```bash
CampusX is the best channel for Data-Science students.
```

**Example 2:**

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:

```bash
CampusX best for DS students.
```

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
# write your code here

test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]

result = []

for i in range(0,len(test_list),2):

    d = {
        key_list[0]: test_list[i],
        key_list[1]: test_list[i+1]
    }

    result.append(d)

print(result)
````

````python
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]

result = []

for i in range(0,len(test_list),2):

    d = {}

    d[key_list[0]] = test_list[i]
    d[key_list[1]] = test_list[i+1]

    result.append(d)

print(result)
````

````python
# when length of key_list is greater than 2

test_list = ["Ali", 10, "Karachi", "Ahmed", 20, "Lahore"]
key_list = ["name", "id", "city"]

result = []

for i in range(0,len(test_list),len(key_list)):

    d = {}

    for key, value in zip(key_list,test_list[i:i + len(key_list)]):
        d[key] = value

    result.append(d)

print(result)
````

### `Q4`: Convert a list of Tuples into Dictionary.

**Example 1:**

Input:
```bash
[("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
```

Output:
```bash
{'akash': [10], 'gaurav': [12], 'anand': [14], 'suraj': [20], 'akhil': [25], 'ashish': [30]}
```

**Example 2:**

Input:
```bash
[('A', 1), ('B', 2), ('C', 3)]
```

Output:
```bash
{'A': [1], 'B': [2], 'C': [3]}
```

````python
# write your code here

lst = [("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]

result = {}

for key, value in lst:
    if key in result:
        result[key].append(value)
    else:
        result[key] = [value]

print(result)
````

### `Q5`: Sort Dictionary key and values List.

**Example 1:**

Input:

```bash
{'c': [3], 'b': [12, 10], 'a': [19, 4]}
```

Output:

```bash
{'a': [4, 19], 'b': [10, 12], 'c': [3]}
```

**Example 2:**

Input:

```bash
{'c': [10, 34, 3]}
```

Output:

```bash
{'c': [3, 10, 34]}
```

````python
# write your code here

d = {'c': [3], 'b': [12, 10], 'a': [19, 4]}

result = dict(sorted(d.items()))

for key in result:
    result[key] = sorted(result[key])

print(result)
````
