### `Q-4`: Number game program.

Write a number game program. Ask the user to enter a number. If the number is greater than number to be guessed, raise a **ValueTooLarge** exception. If the value is smaller the number to be guessed the, raise a **ValueTooSmall** exception and prompt the user to enter again. Quit the program only when the user enters the correct number. Also raise **GuessError** if user guess a number less than 1.

````python
# Write code here

import random

class ValueTooLarge(Exception):
    pass

class ValueTooSmall(Exception):
    pass

class GuessError(Exception):
    pass

num = random.randint(1,100)

while True:
    try:

        guess = int(input('Enter a number:'))

        if guess < 1:
            raise GuessError('Guess must be greater than or equal to 1')

        if guess > num:
            raise ValueTooLarge('Your guess is too large')

        if guess < num:
            raise ValueTooSmall('Your guess is too small')

        print('Correct guess')
        break

    except GuessError as e:
        print(e)

    except ValueTooLarge as e:
        print(e)

    except ValueTooSmall as e:
        print(e)
````

### `Q-5:` Cast vote

To handle the age, create **InvalidAge** exception and for name, create **InvalidName** exception. The name will be invalid when the string will be empty or name has only one word.

Example 1:

```bash

```

Output:

```bash
Goransh Singh  Congratulation !!! You can vote.
```

````python
# Write code here

class InvalidName(Exception):
    pass
class InvalidAge(Exception):
    pass

try:
    name = input('Enter name:')

    word_count = len(name.split())

    # if name == "" or word_count < 2:
      #  raise InvalidName('Invalid name: name must contain at least two words')

    if word_count < 2:    # best approach
        raise InvalidName('Invalid name: name must contain at least two words')

    age = int(input('Enter age:'))

    if age < 18:
        raise InvalidAge('Invalid age: age must be 18 or above')

    print('Congratulation!!!',name.title(),'You can vote')

except InvalidName as e:
    print(e)

except InvalidAge as e:
    print(e)
````

### `Q-6`: Write a python function which infinitely prints natural numbers in a single line. Raise the **StopIteration** exception after displaying first 20 numnbers to exit from the program.

````python
# Write code here

def print_natural():
    i = 1
    while True:
        print(i,end = ' ')
        i += 1
        if i == 21:
            raise StopIteration('First 20 Natural numbers have been printed')
try:
    print_natural()
except StopIteration as e:
    print('\n',e)
````
