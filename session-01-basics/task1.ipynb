### Q2:- Write a program that will convert celsius value to fahrenheit.

The formula to convert Celsius (°C) into Fahrenheit (°F) is:

F = C * (9/5) + 32
	​


````python
# Write your code here
Celsius = float(input('Enter temperature in celsius: '))

Fahrenheit = (9/5*Celsius + 32)
print('Temperature in Fahrenheit: ',Fahrenheit)
````

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.



````python
# Write your code here

import math
# import is a keyword used to bring a module into your program.
# math is a built-in Python module containing mathematical functions.

x1 = float(input('Enter x1: '))
y1 = float(input('Enter y1: '))
x2 = float(input('Enter x2: '))
y2 = float(input('Enter y2: '))

d = ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5
# d = ((x2 - x1)**2 + (y2 - y1)**2) ** (1/2)
# d = math.sqrt((x2-x1)**2 + (y2-y1)**2)

print('The Euclidean distance between two coordinates is',d)
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Write your code here

# The formula for Simple Interest (SI) is:
# SI = (P*R*T)/100
# SI = Simple Interest
# P = Principal amount (original money)
# R = Rate of interest (per year)
# T = Time (in years)

P = float(input('Enter the value of principle: '))
R = float(input('Enter rate of interest(per year): '))
T = float(input('Enter time(in years)'))
SI = (P*R*T)/100

print('The Simple Interest is:',SI)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here

heads = int(input("Enter heads: "))
legs = int(input("Enter legs: "))

dogs = (legs - 2 * heads) // 2
chickens = heads - dogs

print("Dogs =", dogs)
print("Chickens =", chickens)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here

n = int(input('Enter the value of n: '))

sum_of_squares = n * (n+1)* (2*n+1) / 6
print('The sum of squares of first n natural numbers is',sum_of_squares)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here

n = int(input('Enter the number of n: '))
a = int(input('Enter 1st term: '))
b = int(input('Enter 2nd term: '))

d = b-a
nth_term = a + (n-1) * d
print('The nth term is: ',nth_term)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here

import math

tank_height = int(input("Enter tank height: "))
tank_length = int(input("Enter tank length: "))
tank_breadth = int(input("Enter tank breadth: "))

glass_height = int(input("Enter glass height: "))
glass_radius = int(input("Enter glass radius: "))

tank_volume = tank_height * tank_length * tank_breadth

glass_volume = math.pi * glass_radius ** 2 * glass_height

number_of_glasses = int(tank_volume // glass_volume)

print("Number of completely filled glasses =", number_of_glasses)
````
