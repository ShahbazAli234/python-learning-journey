###`Q-2:` Read through a text file, line by line. Use a dict to keep track of how many times each vowel (a, e, i, o, and u) appears in the file. Print the resulting tabulation -- dictionary.

````python
# Write code here

def count_vowels(filename):
    with open(filename, 'r') as f:

        vowels = {
            'a':0,
            'e':0,
            'i':0,
            'o':0,
            'u':0
        }

        for line in f:
            for char in line.lower():
                if char in vowels:
                    vowels[char] += 1

    return vowels

count_vowels('Sample')
````

In last add a line Total, by summing the value of third column

Output File Example:

````python
# write code here

with open('Sample1', 'r') as f:
    with open('output.txt', 'w') as out:

        total =  0

        for line in f:
            numbers = line.split('\t')
            first = int(numbers[0])
            second = int(numbers[1])

            result = first * second
            out.write(f"{first}\t{second}\t{result}\n")
            total += result

        out.write(f"Total\t{total}")
````

###`Q-4:` Create line wise reverse of a file
Write a function which takes two arguments: the names of the input file (to be read from) and the output file (which will be created).

For example, if a file looks like
 ```
abc def
ghi jkl
```
then the output file will be
```
fed cba
lkj ihg
```
**Notice**: The newline remains at the end of the string, while the rest of the characters are all reversed.

````python
with open('input', 'w') as f:
    f.write('abc def\n')
    f.write('ghi jkl')
````

````python
# write code here

def reverse(inputfilename,outputfilename):
    with open(inputfilename,'r') as f:
        with open(outputfilename, 'w') as out:
            for line in f:
                lin = line.strip()
                new_line = lin[::-1]
                new_line  = new_line + '\n'
                out.write(new_line)
````

````python
reverse('input', 'output')
````

````python
# CODE 2

def reverse(inputfilename, outputfilename):
    with open(inputfilename, 'r') as f:
        with open(outputfilename, 'w') as out:

            for line in f:
                if line.endswith('\n'):
                    line = line[:-1]
                    out.write(line[::-1] + '\n')
                else:
                    out.write(line[::-1])


reverse('input', 'output')
````

````python
# write code here

import pickle
import string

strings = """Alice was beginning to get very tired of sitting by her sister
on the bank, and of having nothing to do: once or twice she had
peeped into the book her sister was reading, but it had no
pictures or conversations in it, `and what is the use of a book,'
thought Alice `without pictures or conversation?'

So she was considering in her own mind (as well as she could,
for the hot day made her feel very sleepy and stupid), whether
the pleasure of making a daisy-chain would be worth the trouble
of getting up and picking the daisies, when suddenly a White
Rabbit with pink eyes ran close by her.

There was nothing so VERY remarkable in that; nor did Alice
think it so VERY much out of the way to hear the Rabbit say to
itself, `Oh dear! Oh dear! I shall be late!' (when she thought
it over afterwards, it occurred to her that she ought to have
wondered at this, but at the time it all seemed quite natural);
but when the Rabbit actually TOOK A WATCH OUT OF ITS WAISTCOAT-
POCKET, and looked at it, and then hurried on, Alice started to
her feet, for it flashed across her mind that she had never
before seen a rabbit with either a waistcoat-pocket, or a watch to
take out of it, and burning with curiosity, she ran across the
field after it, and fortunately was just in time to see it pop
down a large rabbit-hole under the hedge."""

word_list = ['alice', 'wonder', 'natural']


# Step 1: Create frequency dictionary
frequency = {}

for word in strings.split():

    # Convert word to lowercase
    word = word.lower()

    # Remove punctuation
    word = word.strip(string.punctuation)

    # Count word
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1


# Step 2: Serialize dictionary
with open('word_frequency.pkl', 'wb') as file:
    pickle.dump(frequency, file)


# Step 3: Deserialize dictionary
with open('word_frequency.pkl', 'rb') as file:
    deserialized_frequency = pickle.load(file)


# Step 4: Show frequency of given words
for word in word_list:
    count = deserialized_frequency.get(word.lower(), 0)
    print(f"{word}: {count}")
````
