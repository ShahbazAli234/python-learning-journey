###`Problem 1:` Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

````python
# Write code here

list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

L = []

for i in range(len(list1)):
    temp = []
    temp.append(list1[i])
    temp.append(list2[i])
    L.append(temp)

print(L)
````

````python
# when both lists have different length

list1 = ["M", "na", "i", "Kh", "X"]
list2 = ["y", "me", "s"]

L = []

if len(list1) < len(list2):
    for i in range(len(list1)):
        L.append([list1[i], list2[i]])
    for i in range(len(list1),len(list2)):
        L.append(list2[i])
else:
    for i in range(len(list2)):
        L.append([list1[i], list2[i]])
    for i in range(len(list2),len(list1)):
        L.append(list1[i])

print(L)
````

###`Problem 3:` Update no of items available

Suppose you are given a list of candy and another list of same size representing no of items of respective candy.

i.e -  
```
candy_list = ['Jelly Belly','Kit Kat','Double Bubble','Milky Way','Three Musketeers']
no_of_items = [10,20,34,74,32]
```

Write a program to show no. of items of each candy type.

`Output:`

```
Jelly Belly-10
Kit Kat-20
Double Bubble-34
Milky Way-74
Three Musketeers-32

```

````python
# Write code here

candy_list = ['Jelly Belly','Kit Kat','Double Bubble','Milky Way','Three Musketeers']
no_of_items = [10,20,34,74,32]

for i in range(len(candy_list)):
    print(candy_list[i],no_of_items[i],sep = '-')

# using zip() function

for candy, items in zip(candy_list, no_of_items):
    print(candy, items, sep='-')
````

###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
# Write code here

L = [2,4,6,10,1]

result = []

for i in L:
    total = 0
    for j in L:
        if j >= i:
            total = total + j
    result.append(total)

print(result)
````

###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
# Write code here

num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]

L = []

for i in range(len(num1)):
    for j in range(len(num2)):
        if num1[i] == num2[j]:
            L.append(num1[i])

L.sort()
print(L)
````

````python
# Without Indexing, directly using the values

num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]

result = []

for i in num1:
    for j in num2:
        if i == j:
            result.append(i)

result.sort()
result
````

###`Problem 7:` Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

`Input:`
```
['1ac21', '23fg', '456', '098d','1','kls']
```

`Output:`
```
['456', '23fg', '1ac21', '1', 'kls', '098d']

```

````python
# Write code here

L = ['1ac21', '23fg', '456', '098d', '1', 'kls']

products = []

# Find the product of digits for each string
for s in L:
    product = 1

    for ch in s:
        if ch.isdigit():
            product = product * int(ch)

    products.append(product)

print(products)
# Sort both lists in descending order
for i in range(len(products)):
    for j in range(i + 1, len(products)):
        if products[i] < products[j]:
            # Swap products
            products[i], products[j] = products[j], products[i]

            # Swap corresponding strings
            L[i], L[j] = L[j], L[i]

print(products)
print(L)
````

````python
lst = ['1ac21', '23fg', '456', '098d', '1', 'kls']

def product_of_digits(s):
    product = 1

    for char in s:
        if char.isdigit():
            product *= int(char)

    return product

result = sorted(lst, key=product_of_digits, reverse=True)

print(result)
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

````python
# Write Code here

L = ['campusxIs', 'bestFor', 'dataScientist']

result = []

for word in L:
    string = ""
    for ch in word:
        if ch.isupper():
            string = string + ' ' + ch
        else:
            string = string + ch

    result.append(string)

print(result)
````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:
```bash
[1,2,3,4,5,7,8]
```

````python
# Write code here

L1 = [1,2,3,4,5,1]
L2 = [2,3,5,7,8]

result = []

for i in L1:
    if i not in result:
        result.append(i)

for i in L2:
    if i not in result:
        result.append(i)

print(result)
````

### `Problem 12:` Write a program that can find the max number of each row of a matrix

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

### `Problem 14:` Write a list comprehension that can transpose a given matrix

matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

[1, 4, 7]<br>
[2, 5, 8]<br>
[3, 6, 9]<br>

````python
# Write code here

# using loop

matrix = [
[1,2,3],
[4,5,6],
[7,8,9]
]

transpose = []

for colomn in range(len(matrix)):
    temp = []

    for row in matrix:
        temp.append(row[colomn])

    transpose.append(temp)

print(transpose)
````

````python
# Write code here

# using list comprehension

matrix = [
[1,2,3],
[4,5,6],
[7,8,9]
]

result = [[row[coloumn] for row in matrix] for coloumn in range(len(matrix))]
print(result)
````

### `Problem 15:` Write a list comprehension that can flatten a nested list

Input<br>
matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

Output:<br>
[1, 2, 3, 4, 5, 6, 7, 8, 9]

````python
# Write code here

matrix = [
[1,2,3],
[4,5,6],
[7,8,9]
]

result = []

for row in matrix:
    for i in row:
        result.append(i)

print(result)
````
