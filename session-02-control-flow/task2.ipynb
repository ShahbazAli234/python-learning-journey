### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here

CTC = int(input('Enter CTC: '))

HRA = 0.1 * CTC
DA =  0.05 * CTC
PF =  0.03 * CTC

salary_after_deductions = CTC - HRA - DA - PF

if CTC < 500000:
    tax = 0
elif CTC < 1000000:
    tax = 0.1 * CTC
elif CTC <= 2000000:
    tax = 0.2 * CTC
else:
    tax = 0.3 * CTC

salary_after_tax = salary_after_deductions - tax
monthly_salary = salary_after_tax/12

print('Monthly In hand Salary is:',monthly_salary)
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# Write code here (10/10)

n = int(input('Enter n: '))
fact = 1

if n >= 0:
    for i in range(1,n+1):
        fact = fact * i
    print('The factorial of',n,'is',fact)

else:
    print('factorial is not possible')
````

````python
# Write code here (10/9.5)
n = int(input('Enter a number: '))
fact = 1
if n > 1:
    for i in range(n,0,-1):
        fact = fact * i
    print('The factorial is',fact)
elif n == 0 or n == 1:
    print('The factorial is 1')
else:
    print('factorial is not possible')
````

### `Problem 7` - Reverse a given integer number.

Example:

`Input:`

```bash
76542
```

`Output:`

```bash
24567
```

````python
# Write code here

n = int(input('Enter a number: '))

original = n

rev = 0

while n != 0:
    digit = n % 10
    n //= 10
    rev = rev * 10 + digit

print('The reversed number is',rev)

if original == rev:
    print('The number is palindrome')
else:
    print('The number is not palindrome')
````

### `Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

**Example 1:**

`Input:`

```bash
30
```

`Output:`

```bash
276
```

````python
# wite code here

n = int(input('Enter N: '))

i = 1
total = 0

while i <= n:

    if i % 5 == 0:
        i += 1
        continue

    if total + i > 300:
        break

    total += i
    i += 1

print('The sum is', total)
````

### `Problem 9`: Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

````python
# Write code here

total = 0
count = 0

while True:
    num = int(input('Enter a number: '))

    if num == 0:
        break

    total += num
    count += 1

if count > 0:
    average = total / count

    print('Sum =', total)
    print('Average =', average)
else:
    print('No numbers were entered.')
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
# Write code here

final_x = 0
final_y = 0

while True:

    movement = input()
    if movement == '!':
        break
    a = movement.split()
    direction = a[0]
    steps =  a[1]
    if direction == 'UP':
        final_y = final_y + int(steps)
    elif direction == 'DOWN':
        final_y = final_y - int(steps)
    elif direction == 'LEFT':
        final_x = final_x - int(steps)
    elif direction == 'RIGHT':
        final_x = final_x + int(steps)

start_x = 0
start_y = 0

d = ((final_x - start_x)**2 + (final_y - start_y)**2)**0.5

print('Distance is',round(d))
````

###`Problem 12`:Write a program to print whether a given number is a prime number or not

````python
n = int(input("Enter a number: "))

if n < 2:
    print("The number is not prime")
else:
    is_prime = True # It is also a flag variable

    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print("The number is prime")
    else:
        print("The number is not prime")
````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
# Write code here

start = int(input('Enter starting number of range: '))
end = int(input('Enter ending number of range: '))

for i in range(start,end+1):
    original = i
    total = 0

    while i != 0:
        digit = i % 10
        total = total + digit * digit * digit
        i = i // 10

    if total == original:
        print(original,end = ' ')
````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
# Write code here

H = int(input('Enter hours: '))
M = int(input('Enter Minutes: '))

hour_hand_angle = (H * 30) + (M * 0.5)
minute_hand_angle = M * 6

difference = hour_hand_angle - minute_hand_angle

angle_1 = abs(difference)
angle_2 = 360 - angle_1

angle = min(angle_1,angle_2)

print('The minimum angle between hour hand and minute hand is',round(angle))
````

###`Problem 15`:Given two rectangles, find if the given two rectangles overlap or not. A rectangle is denoted by providing the x and y coordinates of two points: the left top corner and the right bottom corner of the rectangle. Two rectangles sharing a side are considered overlapping. (L1 and R1 are the extreme points of the first rectangle and L2 and R2 are the extreme points of the second rectangle).

Note: It may be assumed that the rectangles are parallel to the coordinate axis.

<img src='https://www.geeksforgeeks.org/wp-content/uploads/rectanglesOverlap.png' width='300' height='200'>

````python
# Write code here

L1_x, L1_y = map(int, input('Enter L1 coordinates (x y): ').split())
R1_x, R1_y = map(int, input('Enter R1 coordinates (x y): ').split())

L2_x, L2_y = map(int, input('Enter L2 coordinates (x y): ').split())
R2_x, R2_y = map(int, input('Enter R2 coordinates (x y): ').split())


if (R1_x < L2_x or
    R2_x < L1_x or
    R1_y > L2_y or
    R2_y > L1_y):

    print('Rectangles do not overlap')

else:
    print('Rectangles overlap')
````
