### **`Problem-1:`** Write a Python function that takes a list and returns a new list with unique elements of the first list.

**Exercise 1:**

Input:

```bash
[1,2,3,3,3,3,4,5]
```

Output:

```bash
[1, 2, 3, 4, 5]
```

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
# Write code here

s = 'green-red-yellow-black-white'

def sort_sequence(s):
    L = s.split('-')
    L.sort()
    return '-'.join(L)

sort_sequence(s)
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# write code here

s = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'

def count_upper_lower(s):
    upper = 0
    lower = 0

    for ch in s:
        if ch.islower():
            lower += 1
        elif ch.isupper():
            upper += 1
    return upper, lower

upper, lower = count_upper_lower(s)

print('No of Upper case characters: ',upper)
print('No of lower case characters: ',lower)
````

````python
# Method 2

def count_case(s):
    upper = 0
    lower = 0

    for ch in s:
        if ch.islower():
            lower += 1
        elif ch.isupper():
            upper += 1

    return f"No. of Upper case characters: {upper}\nNo. of Lower case characters: {lower}"
    # f"..." (f-string)
    # The f before the quotes means formatted string literal (or simply f-string).

print(count_case(s))
````

### **`Problem 4:`** Write a Python program to print the even numbers from a given list.

```
Sample List : [1, 2, 3, 4, 5, 6, 7, 8, 9]

```

````python
# Write code here

L = [1, 2, 3, 4, 5, 6, 7, 8, 9]

def is_even(L):

    result = []

    for i in L:
        if i % 2 == 0:
            result.append(i)

    return result

print(is_even(L))
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here

def check_perfect(num):

    total = 0

    for i in range(1,num):
        if num % i == 0:
            total += i

    if total == num:
        return 'perfect number'
    else:
        return 'Not perfect number'

check_perfect(8128)
````

````python
# Write code here

def check_perfect(n):

    total = 0

    for i in range(1,n+1):
        if n % i == 0:
            total += i

    if total == n * 2:
        return 'This is a perfect number'
    else:
        return 'This is not a perfect number'

print(check_perfect(8128))
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
# Write code here

dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}

def concatenate_dict(*args):

    result = {}

    for dic in args:
        result.update(dic)

    return result

concatenate_dict(dic1, dic2, dic3)
````

`Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
# Write code here

s = 'hello how are you i am fine thank you'

def most_frequent_words(s):

    d = {}

    words = s.split()

    for word in words:
        if word not in d:
            d[word] = 1
        else:
            d[word] += 1

    max_count = 1
    max_word = words[0]

    for (key,value) in d.items():
        if value > max_count:
            max_count = value
            max_word = key

    return max_word, max_count

max_word, max_count = most_frequent_words(s)

print(max_word,'-->',max_count)
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
# Write code here

L = [13,42,15,37,22,39,41,50]

def hstogrm(L):

    d = {}

    for i in L:
        if i / 10 != 0:
            start = (i // 10) * 10  + 1
            end = start + 9
        else:
            start = i - 9
            end = i

        bin = f"{start}-{end}"

        if bin not in d:
            d[bin] = 1
        else:
            d[bin] += 1

    return d

hstogrm(L)
````

`Problem-9` Write a python function that accepts a list of 2D co-ordinates and a query point, and then finds the the co-ordinate which is closest in terms of distance from the query point.

```
List of Coordinates
[(1,1),(2,2),(3,3),(4,4)]
Query Point
(0,0)
```

```
Output
Nearest to (0,0) is (1,1)
```

````python
# Write code here

points = [(1,1),(2,2),(3,3),(4,4)]
Query_Point = (0,0)

def find_coordinate(points, query_point):

    nearest = points[0]

    min_distance = (points[0][0] - Query_Point[0])**2 + (points[0][1] - Query_Point[1])**2

    for point in points:
        distance = (point[0] - Query_Point[0])**2 + (point[1] - Query_Point[1])**2
        if distance < min_distance:
            min_distance = distance
            nearest = point

    return nearest

nearest = find_coordinate(points,Query_Point)

print('Nearest to',Query_Point,'is',nearest)
````

`Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
# Write code here

strings = [
    "I love Python",
    "Python is easy",
    "I love coding"
]

def bag_of_word(strings):
    unique_words = set()
    for sentence in strings:
        words = sentence.split()

        for word in words:
            unique_words.add(word)

    unique_words = sorted(unique_words)

    result = []

    for sentence in strings:
        words = sentence.split()
        row = []

        for word in unique_words:
           row.append(words.count(word))

        result.append(row)
    print("Vocabulary:", unique_words)
    print()

    for i in range(len(strings)):
        print(strings[i])
        print(result[i])
        print()

bag_of_word(strings)
````

###`Problem-12:`Write a Python program to create a list containing the power of said number in bases raised to the corresponding number in the index using Python map.
`Input:`
```
list1 = [1,2,3,4,5,6]
```
`Output:`
```
[1,2,9,64,625,-]
```


````python
# Write code here

list1 = [1,2,3,4,5,6]

result = list(map(lambda x:x[1] ** x[0],enumerate(list1)))
print(result)
````

````python
# METHOD 2 --> Not used lambda and map

list1 = [1,2,3,4,5,6]

L = []

for i,x in enumerate(list1):
    L.append(x**i)

print(L)
````

````python
# using list comprehension
result = [value ** index for index, value in enumerate(list1)]
print(result)
````
